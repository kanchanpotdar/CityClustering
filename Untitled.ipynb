```python
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt

def load_image(image_path):
    """Load the image and convert it to a grayscale NumPy array."""
    image = Image.open(image_path)
    gray_image = image.convert('L')
    return np.array(gray_image)

def find_illuminated_pixels(image_array, threshold=200):
    """Find the coordinates of illuminated pixels (white dots) in the image."""
    illuminated_pixels = np.argwhere(image_array > threshold)  # Assuming white pixels have intensity > threshold
    return illuminated_pixels

def euclidean_distance(a, b):
    """Compute the Euclidean distance between two points a and b."""
    return np.sqrt(np.sum((a - b)**2))

def k_means_clustering(data, k, max_iterations=100):
    """Implement the K-Means algorithm."""
    # Randomly initialize centroids
    centroids = data[np.random.choice(data.shape[0], k, replace=False)]

    for _ in range(max_iterations):
        # Assign each data point to the nearest centroid
        distances = np.array([[euclidean_distance(point, centroid) for centroid in centroids] for point in data])
        cluster_assignment = np.argmin(distances, axis=1)
      
        """Update centroids by calculating the mean of the points in each cluster."""
        new_centroids = np.zeros((k, data.shape[1]))  # Initialize a new array for centroids
        for i in range(k):
            points_in_cluster = data[cluster_assignment == i]  # Get points in cluster i
            if len(points_in_cluster) > 0:  # Check to avoid division by zero
                new_centroids[i] = points_in_cluster.mean(axis=0)  # Calculate mean and update centroid
    
        # Check for convergence
        if np.all(centroids == new_centroids):
            break

        centroids = new_centroids

    return centroids, cluster_assignment

def calculate_wcss(data, centroids, cluster_assignment):
    """Calculate the within-cluster sum of squares (WCSS)."""
    wcss = 0
    for i, centroid in enumerate(centroids):
        cluster_points = data[cluster_assignment == i]
        wcss += np.sum([euclidean_distance(point, centroid)**2 for point in cluster_points])
    return wcss

def elbow_method(data, max_clusters):
    """Use the Elbow Method to find the optimal number of clusters."""
    wcss = []

    for k in range(1, max_clusters + 1):
        centroids, cluster_assignment = k_means_clustering(data, k)
        wcss_value = calculate_wcss(data, centroids, cluster_assignment)
        wcss.append(wcss_value)

    plt.plot(range(1, max_clusters + 1), wcss, marker='o')
    plt.xlabel('Number of clusters (K)')
    plt.ylabel('Within-cluster sum of squares (WCSS)')
    plt.title('Elbow Method')
    plt.show()

    # Determine the elbow point
    diff = np.diff(wcss)
    diff2 = np.diff(diff)
    elbow_point = np.argmax(diff2) + 2  # +2 because diff2 is second derivative and indexing starts from 0

    return elbow_point

def calculate_distances(centroids):
    """Calculate distances between centroids (city locations)."""
    num_cities = len(centroids)
    distances = np.zeros((num_cities, num_cities))
    for i in range(num_cities):
        for j in range(i+1, num_cities):
            distances[i, j] = euclidean_distance(centroids[i], centroids[j])
            distances[j, i] = distances[i, j]
    return distances

def plot_clusters(image_array, centroids, cluster_assignment):
    """Plot the clusters on the image."""
    plt.figure(figsize=(8, 8))
    plt.imshow(image_array, cmap='gray')
    plt.scatter(centroids[:, 1], centroids[:, 0], marker='x', color='red', s=100, label='City Centers')
    plt.legend()
    plt.title('City Locations Identified ')
    plt.show()

# Main script logic
image_paths = {"1.png","2.png","3.png","4.png"}
max_clusters = 10  # Maximum number of clusters to consider
for image_path in image_paths:
    # Step 1: Load the image and convert it to a 2D NumPy array
    image_array = load_image(image_path)

    # Step 2: Find the illuminated pixels (white dots)
    illuminated_pixels = find_illuminated_pixels(image_array)

    # Step 3: Use the Elbow Method to determine the optimal number of clusters
    optimal_clusters = elbow_method(illuminated_pixels, max_clusters)
    print(f'Optimal number of clusters: {optimal_clusters}')

    # Step 4: Perform K-Means clustering with the optimal number of clusters
    centroids, cluster_assignment = k_means_clustering(illuminated_pixels, optimal_clusters)

    # Print the coordinates of the centroids
    print("Coordinates of centroids (city centers):")
    for i, centroid in enumerate(centroids):
        print(f'City {i+1}: {centroid}')

    # Step 5: Calculate distances between city locations
    distances = calculate_distances(centroids)
    print("Distances between cities:")
    print(distances)

    # Step 6: Plot clusters on the image
    plot_clusters(image_array, centroids, cluster_assignment)
```

```python
import numpy as np
from PIL import Image
import random
import matplotlib.pyplot as plt

# Function to load and preprocess the image
def load_image(image_path):
    image = Image.open(image_path).convert('L')  # Convert to grayscale
    return np.array(image)

# Function to identify the light points
def find_illuminated_pixels(image_array, threshold=200):
    return np.argwhere(image_array > threshold)

# Function to calculate Euclidean distance
def euclidean_distance(point1, point2):
    return np.linalg.norm(point1 - point2)

# Function to initialize centroids randomly
def initialize_centroids(points, k):
    return points[random.sample(range(points.shape[0]), k)]

# Function to assign points to clusters based on nearest centroid
def assign_clusters(points, centroids):
    clusters = [[] for _ in centroids]
    for point in points:
        distances = [euclidean_distance(point, centroid) for centroid in centroids]
        clusters[np.argmin(distances)].append(point)
    return clusters

# Function to compute centroids as the mean of points in each cluster
def compute_centroids(clusters):
    return [np.mean(cluster, axis=0) for cluster in clusters if len(cluster) > 0]

# K-Means clustering algorithm
def kmeans(points, k, max_iterations=100, tol=1e-4):
    centroids = initialize_centroids(points, k)
    for _ in range(max_iterations):
        clusters = assign_clusters(points, centroids)
        new_centroids = compute_centroids(clusters)
        
        # Ensure the number of centroids remains the same
        while len(new_centroids) < k:
            missing_centroids = k - len(new_centroids)
            new_centroids.extend(initialize_centroids(points, missing_centroids))
        
        if np.all(np.abs(np.array(new_centroids) - np.array(centroids)) < tol):
            break
        centroids = new_centroids
    return centroids, clusters

# Custom silhouette score calculation
def silhouette_score(points, labels):
    unique_labels = np.unique(labels)
    silhouette_scores = []
    
    for idx, point in enumerate(points):
        same_cluster_points = points[labels == labels[idx]]
        other_cluster_points = points[labels != labels[idx]]
        
        if len(same_cluster_points) > 1:
            a = np.mean([euclidean_distance(point, other_point) for other_point in same_cluster_points if not np.array_equal(point, other_point)])
        else:
            a = 0
        
        b = np.inf
        for label in unique_labels:
            if label != labels[idx]:
                cluster_points = points[labels == label]
                distance = np.mean([euclidean_distance(point, other_point) for other_point in cluster_points])
                if distance < b:
                    b = distance
        
        silhouette_scores.append((b - a) / max(a, b))
    
    return np.mean(silhouette_scores)

# Function to determine the optimal number of clusters using the silhouette method
def find_optimal_k(points, max_k=10):
    best_k = 2
    best_score = -1
    for k in range(2, max_k + 1):
        centroids, clusters = kmeans(points, k)
        labels = np.zeros(points.shape[0])
        for cluster_id, cluster in enumerate(clusters):
            for point in cluster:
                labels[np.all(points == point, axis=1)] = cluster_id
        score = silhouette_score(points, labels)
        if score > best_score:
            best_k = k
            best_score = score
    return best_k

# Function to calculate distances between clusters manually
def calculate_distances(centroids):
    num_centroids = len(centroids)
    distances = np.zeros((num_centroids, num_centroids))
    for i in range(num_centroids):
        for j in range(i + 1, num_centroids):
            distance = euclidean_distance(centroids[i], centroids[j])
            distances[i, j] = distance
            distances[j, i] = distance
    return distances

# Function to visualize the image with cluster centers
def show_image_with_centroids(image_array, centroids):
    plt.imshow(image_array, cmap='gray')
    for i, centroid in enumerate(centroids):
        plt.scatter(centroid[1], centroid[0], c='red', marker='x')
    plt.title('Image with Cluster Centers')
    plt.show()

# Function to print coordinates of each cluster
def print_cluster_coordinates(centroids):
    for i, centroid in enumerate(centroids):
        print(f"City {i+1}: {centroid}")

# Loop to process multiple images
image_paths = ["1.png", "2.png", "3.png", "4.png"]
max_clusters = 10

for image_path in image_paths:
    print(f"Processing image: {image_path}")
    
    # Step 1: Load and preprocess the image
    image_array = load_image(image_path)
    
    # Step 2: Identify the light points
    light_points = find_illuminated_pixels(image_array)
    
    # Step 3: Determine the optimal number of clusters using the silhouette method
    optimal_k = find_optimal_k(light_points, max_k=max_clusters)
    print("Optimal K:", optimal_k)
    
    # Step 4: Perform K-Means clustering with the optimal number of clusters
    centroids, clusters = kmeans(light_points, optimal_k)
    
    # Step 5: Calculate distances between clusters manually
    distances = calculate_distances(centroids)
    print("Pairwise distances between centroids:\n", distances)
    
    # Step 6: Visualize the image with cluster centers
    show_image_with_centroids(image_array, centroids)
    
    # Step 7: Print coordinates of each cluster
    print_cluster_coordinates(centroids)
```
